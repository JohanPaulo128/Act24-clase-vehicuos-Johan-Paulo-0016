###4. **Vehiculo**
   - Atributos: marca, modelo, año, kilometraje
   - Funciones: encender(), acelerar(), frenar()

![texto del vínculo](https://raw.githubusercontent.com/JohanPaulo128/Act24-clase-vehicuos-Johan-Paulo-0016/refs/heads/main/infoclaseJohan0016.png)

```python
class Vehiculo:
    # Método constructor para inicializar atributos
    def __init__(self, marca, modelo, anio, kilometraje):
        self.marca = marca
        self.modelo = modelo
        self.anio = anio
        self.kilometraje = kilometraje
        self.encendido = False

    # Definición de funciones (Métodos)
    def encender(self):
        self.encendido = True
        return f"🔥 El {self.marca} {self.modelo} ha sido encendido. ¡Listo para arrancar!"

    def acelerar(self):
        if self.encendido:
            return f"🚀 El {self.marca} está acelerando. Kilometraje actual: {self.kilometraje} km."
        else:
            return f"❌ No puedes acelerar, el {self.marca} está apagado."

    def frenar(self):
        return f"🛑 El {self.marca} {self.modelo} ha frenado totalmente."

# --- Sección Interactiva ---
print("--- Registro de Vehículos ---")

# Creación del Objeto 1 con datos del usuario
print("\nDatos para el Vehículo 1:")
v1_marca = input("Marca: ")
v1_modelo = input("Modelo: ")
v1_anio = int(input("Año: "))
v1_km = int(input("Kilometraje: "))

vehiculo1 = Vehiculo(v1_marca, v1_modelo, v1_anio, v1_km)

# Creación del Objeto 2 con datos del usuario
print("\nDatos para el Vehículo 2:")
v2_marca = input("Marca: ")
v2_modelo = input("Modelo: ")
v2_anio = int(input("Año: "))
v2_km = int(input("Kilometraje: "))

vehiculo2 = Vehiculo(v2_marca, v2_modelo, v2_anio, v2_km)

# --- Ejecución de Funciones y Mensajes ---
print("\n" + "="*30)
print("DEMOSTRACIÓN DE FUNCIONES")
print("="*30)

# Pruebas Vehículo 1
print(f"\n[Vehículo 1: {vehiculo1.marca}]")
print(vehiculo1.encender())
print(vehiculo1.acelerar())
print(vehiculo1.frenar())

# Pruebas Vehículo 2
print(f"\n[Vehículo 2: {vehiculo2.marca}]")
print(vehiculo2.encender())
print(vehiculo2.acelerar())
print(vehiculo2.frenar())

print("\n" + "="*30)
print("FIN DEL PROGRAMA")
print("Programa realizado por Johan Paulo NL = 04")
print("="*30)
```
